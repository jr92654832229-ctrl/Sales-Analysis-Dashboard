# **Part D: Data Visualization**

```python
# BAR CHART
import matplotlib.pyplot as plt

# Group data by Month and calculate total sales
monthly_sales = df.groupby("Month")["Sales"].sum().sort_index()

# Create figure
plt.figure(figsize=(12,5))

# Create bar chart
plt.bar(monthly_sales.index, monthly_sales.values)

# Add chart title
plt.title("Monthly Sales")

# Add X-axis label
plt.xlabel("Month")

# Add Y-axis label
plt.ylabel("Total Sales")

# Display chart
plt.show()
```

```python
# LINE CHART
import matplotlib.pyplot as plt

# Calculate monthly sales
monthly_Sales = df.groupby("month")["Sales"].sum().sort_index()

# Create line chart
plt.figure(figsize=(10, 6))
plt.plot(monthly_Sales.index, monthly_Sales.values, marker='o')

plt.title("monthly sales trend")
plt.ylabel("Sales")
plt.xlabel("Month")
plt.grid(True)
```

```python
# PIE CHART
import matplotlib.pyplot as plt

# Calculate total sales for each city
city_sales = df.groupby("City")["Sales"].sum()

plt.figure(figsize=(8,8))

# Calculate total sales for each city
plt.pie(city_sales.values, labels=city_sales.index, autopct="%1.1f%%")

plt.title("Sales by City")

plt.show()
```

```python
# SCATTER CHART
import matplotlib.pyplot as plt

plt.figure(figsize=(10,5))

plt.scatter(df["Quantity Ordered"], df["Sales"])

plt.title("quantity order vs sales")
plt.xlabel("quantity ordered")
plt.ylabel("sales")

plt.show()
```

```python
# HISTOGRAM CHART
import matplotlib.pyplot as plt

plt.figure(figsize=(10,6))

plt.hist(df["Sales"],bins=20)

plt.title("sales distribution")
plt.xlabel("sales")
plt.ylabel("frequency")

plt.show()
```

```python
# BOX PLOT CHART

import matplotlib.pyplot as plt

plt.figure(figsize=(8,6))

plt.boxplot(df["Sales"])

plt.title("sales distribution")
plt.ylabel("Sales")

plt.show()
```

## **Part F: Business Insights**

```python
import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots

# 1. Product-wise Sales (Bar Chart)
category_sales = df.groupby("Product")["Sales"].sum().reset_index()

fig = px.bar(
    category_sales,
    x='Product',
    y='Sales',
    title='sales by category'

)
fig.show()
```

```python
import pandas as pd
import plotly.express as px

df = pd.read_csv('/content/Sales_Data[1].csv')

# 2. City-wise Sales (Bar Chart)
city_sales = df.groupby('City')['Sales'].sum().reset_index()

fig = px.bar(
    city_sales,
    x='City',
    y='Sales',
    title='Sales by City'
)

fig.show()
```

```python
# 3. Monthly Sales Trend (Line Chart)

monthly_sales = df.groupby("Month")["Sales"].sum().reset_index()

fig = px.line(
    monthly_sales,
    x='Month',
    y='Sales',
    title='monthly sales'
)
fig.show()
```

```python
# 4. Product Sales Distribution (Pie Chart)

product_sales = df.groupby('Product')['Sales'].sum().reset_index()
# Pie chart
fig = px.pie(
    product_sales,
    names='Product',
    values='Sales',
    title='Product Sales Distribution'
)

fig.show()
```

```python
# 5. Quantity vs Sales (Scatter Plot)

fig = px.scatter(
    df,
    x='Quantity Ordered',
    y='Sales',
    color='City',
    title='Quantity vs Sales'
)

fig.show()
```
